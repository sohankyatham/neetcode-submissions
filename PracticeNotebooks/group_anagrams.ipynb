```python
# Problem: Group Anagrams 
#
# Brute force O(n^2) Solution
# Output: [["hat"],["act", "cat"],["stop", "pots", "tops"]]

def group_anagrams(strs):
    visited = [False]*len(strs)
    answer = []

    for i in range(len(strs)):
        if visited[i] == True:
            continue

        group = []

        for j in range(i, len(strs)):
            if sorted(strs[i]) == sorted(strs[j]):
                group.append(strs[j])
                visited[j] = True

        answer.append(group)
    
    return answer

group_anagrams(["act","pots","tops","cat","stop","hat"])
```

```python
# Optimized O(n) Solution
# Output: [["hat"],["act", "cat"],["stop", "pots", "tops"]]

def group_anagrams(strs):
    ourMap = {}

    for i in range(len(strs)):
        sorted_key = "".join(sorted(strs[i]))

        if sorted_key in ourMap:
            ourMap[sorted_key].append(strs[i]) 
        else:
            ourMap[sorted_key] = [strs[i]]
    
    return list(ourMap.values()) 

group_anagrams(["act","pots","tops","cat","stop","hat"])



# '''
# My initial thoughts for approaching this:

# go thru the list, start with 'act' since its unique add it as a key and also the value in the hashmap, go to 'pots' and thats unique so add
# it as a key and the value, then go to 'tops', but 'tops' sorted is same as 'pots' so add it as a value to the second key and so on.
# To add additional values do "".join(value).

# Then for the answer get all of the values for each of the keys.
# '''
```
