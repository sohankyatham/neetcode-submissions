```python
'''
Input: nums = [1,2,2,3,3,3], k = 2

Output: [2,3]



Input: nums = [7,7], k = 1

Output: [7]
'''
```

```python
# Initial Approach for Solving 

# This problem is essentially find the top K (an integer) recurring elements in a list of numbers

def top_k(nums, k):
    ourMap = {}

    # 1) loop thru the list and add unique elements as keys
    # 2) recurring elements should updatae the count (value)
    for i in range(len(nums)):
        if nums[i] in ourMap:
            ourMap[nums[i]] = ourMap.get(nums[i], 0) + 1    # increment the count
        else:
            ourMap[nums[i]] = 1

    # '''
    # more pythonic way consider this:

    # for num in nums:
    #     ourMap[num] = ourMap.get(num,0) + 1
    # '''

    # 3) return the top k
    sorted_map = sorted(ourMap.items(), key=lambda x: x[1], reverse=True)

    answer = []
    for i in range(k):
        answer.append(sorted_map[i][0])
    
    return answer
    

top_k([1,2,2,2,3,3,3,3,3], 2)
```
