```python
# Problem: Two Sum
# This notebook is for me to redo the problems and write my thoughts & any notes 


# The input is nums = [3,4,5,6], target = 7
# Output: [0, 1], basically find the indices i and j that sum to target like: nums[i] + nums[j] == target


# Brute Force Solution
def two_sum(nums, target): 
    for i in range(len(nums)):  # nested for loops O(n^2)
        for j in range(i+1, len(nums)): # by saying i, you start there, but i+1 allows you to start at the next index after i
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

two_sum([3,4,5,6], 7)   
```

```python
# The O(n) optimized solution
'''
PLAN & THOUGHTS:

basically what we can do is loop through the nums list, put everything in a hash map
to store the data in key-value pairs and use for lookup

In the hashmap, the key would be the number and the value would be its index.

then do target minus an num and then find if there is another num (complement) that can be used
find the indices for those two numbers and that is the solution. map each number to the index and return

This will bring the time complexity down to O(n) and increases space complexity to O(n). 
The brute force solution was O(n^2) time complexity and O(1) space complexity. 

TRADEOFF: Our optimized solution reduces the time complexity and increases the space complexity.

**dont confuse hashmap with hashset, hashmap stores key-value pairs and is used for lookup
hashset is stores unique keys 
'''
```

```python
# FINAL SOLUTION
def two_sum(nums, target):
    ourMap = {}

    for i in range(len(nums)):
        need = target - nums[i]

        if (need in ourMap):
            return [ourMap[need], i]
        
        ourMap[nums[i]] = i

    return []

two_sum([3,4,5,6], 7)
```
